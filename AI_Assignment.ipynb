```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}

start = input("Enter starting node: ")

queue = [start]
visited = []
bfs = []

while queue:
    node = queue.pop(0)

    if node not in visited:
        visited.append(node)
        bfs.append(node)
        queue.extend(graph[node])

visited = []
stack = [start]
dfs = []

while stack:
    node = stack.pop()

    if node not in visited:
        visited.append(node)
        dfs.append(node)
        stack.extend(graph[node][::-1])

print("BFS:", bfs)
print("DFS:", dfs)
```

```python
a = int(input("Enter capacity of Jug A: "))
b = int(input("Enter capacity of Jug B: "))
target = int(input("Enter target amount: "))

queue = [(0, 0)]
visited = [(0, 0)]
parent = {}

while queue:
    x, y = queue.pop(0)

    if x == target or y == target:
        goal = (x, y)
        break

    states = [
        (a, y),
        (x, b),
        (0, y),
        (x, 0)
    ]

    pour = min(x, b - y)
    states.append((x - pour, y + pour))

    pour = min(y, a - x)
    states.append((x + pour, y - pour))

    for state in states:
        if state not in visited:
            visited.append(state)
            queue.append(state)
            parent[state] = (x, y)

path = []
state = goal

while state != (0, 0):
    path.append(state)
    state = parent[state]

path.append((0, 0))
path.reverse()

print("Minimum steps:", len(path) - 1)

for state in path:
    print("Jug A:", state[0], "Jug B:", state[1])
```

```python
graph = {}

n = int(input("Enter number of cities: "))

for i in range(n):
    city = input("Enter city name: ")
    graph[city] = []

m = int(input("Enter number of roads: "))

for i in range(m):
    a, b = input("Enter connected cities: ").split()
    graph[a].append(b)
    graph[b].append(a)

start = input("Enter starting city: ")
end = input("Enter destination city: ")

stack = [start]
visited = []

while stack:
    city = stack.pop()

    if city not in visited:
        visited.append(city)
        stack.extend(graph[city])

if end in visited:
    print("Route exists")
else:
    print("Route does not exist")

visited = []

print("Connected Components:")

for city in graph:
    if city not in visited:
        stack = [city]
        component = []

        while stack:
            current = stack.pop()

            if current not in visited:
                visited.append(current)
                component.append(current)
                stack.extend(graph[current])

        print(component)
```
